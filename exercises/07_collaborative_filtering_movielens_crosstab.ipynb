```python
import pandas as pd
from fastai.collab import *
from fastai.tabular.all import *

# 1. Download and extract the sample MovieLens dataset
path = untar_data(URLs.ML_SAMPLE)

# 2. Read the ratings CSV file
ratings = pd.read_csv(path/'ratings.csv')

# 3. Create a crosstab representation (Users as rows, Movies as columns)
crosstab = ratings.pivot(index='userId', columns='movieId', values='rating')

# 4. View a small subset of the matrix
crosstab.iloc[:10, :10]
```
